```python

import pandas as pd

# Load the credit card transaction dataset
df = pd.read_csv("/content/creditcard.csv")

print("Dataset loaded successfully!")
print("-" * 40)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nFirst 5 transactions:")
display(df.head())
```

```python

# Inspect dataset structure

print("DATASET INFORMATION")
print("=" * 50)

df.info()

print("\nCOLUMN DATA TYPES")
print("=" * 50)
display(df.dtypes.to_frame(name="Data Type"))

print("\nSTATISTICAL SUMMARY")
print("=" * 50)
display(df.describe())
```

```python

# Check missing values in each column

print("MISSING VALUES")
print("=" * 40)

missing_values = df.isnull().sum()

display(missing_values.to_frame(name="Missing Count"))

print("Total missing values:", df.isnull().sum().sum())

# Check duplicate records
print("\nDUPLICATE RECORDS")
print("=" * 40)

duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)
print("Total rows before cleaning:", len(df))
```

```python

# Display some duplicate records

duplicate_rows = df[df.duplicated(keep=False)]

print("Total rows involved in duplication:",
      len(duplicate_rows))

print("\nFirst 10 duplicate records:")
display(duplicate_rows.head(10))

print("\nNumber of duplicate rows to remove:",
      df.duplicated().sum())
```

```python

# Store original row count
rows_before = len(df)

# Remove exact duplicate rows
df = df.drop_duplicates(keep="first")

# Store updated row count
rows_after = len(df)

# Calculate removed rows
rows_removed = rows_before - rows_after

print("DUPLICATE REMOVAL REPORT")
print("=" * 40)
print("Rows before cleaning:", rows_before)
print("Duplicate rows removed:", rows_removed)
print("Rows after cleaning:", rows_after)
print("Remaining duplicates:", df.duplicated().sum())
```

```python

# Check missing values after duplicate removal

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().sum() / len(df)) * 100
})

print("MISSING VALUE REPORT")
print("=" * 45)

display(missing_report)

print("Total missing values:", df.isnull().sum().sum())
```

```python

import numpy as np

print("NumPy imported successfully!")
```

```python

print("DATA VALIDATION REPORT")
print("=" * 45)

# 1. Check data types
print("\nColumn data types:")
print(df.dtypes.value_counts())

# 2. Check transaction classes
print("\nTransaction class values:")
print(df["Class"].value_counts())

# 3. Check invalid class values
print("\nInvalid Class values:",
      (~df["Class"].isin([0, 1])).sum())

# 4. Check infinite values
print("\nInfinite values:",np.isinf(df.select_dtypes(include=np.number)).sum().sum())

# 5. Check negative transaction amounts
print("\nNegative transaction amounts:",
      (df["Amount"] < 0).sum())

# 6. Check negative time values
print("\nNegative time values:",
      (df["Time"] < 0).sum())
```

```python

# OUTLIER DETECTION USING IQR

Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["Amount"] < lower_limit) |
    (df["Amount"] > upper_limit)
]

print("IQR OUTLIER REPORT")
print("=" * 40)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

print("\nPotential outliers:", len(outliers))
print("Percentage of dataset:",
      round(len(outliers) / len(df) * 100, 2), "%")
```

```python

# Inspect potential outliers

print("HIGHEST TRANSACTION AMOUNTS")
display(
    outliers.sort_values(
        by="Amount",
        ascending=False
    )[["Amount", "Class"]].head(10)
)

print("\nOUTLIER CLASS DISTRIBUTION")
print(outliers["Class"].value_counts())

print("\nOUTLIER AMOUNT STATISTICS")
display(outliers["Amount"].describe())
```

```python
print("HIGHEST TRANSACTION AMOUNTS")
display(
    outliers.sort_values(
        by="Amount",
        ascending=False
    )[["Amount", "Class"]].head(10)
)

print("\nOUTLIER CLASS DISTRIBUTION")
print(outliers["Class"].value_counts())

print("\nOUTLIER AMOUNT STATISTICS")
display(outliers["Amount"].describe())
```

```python

# Display the 10 largest flagged transactions
print("TOP 10 POTENTIAL OUTLIERS")

display(
    outliers.sort_values(
        by="Amount",
        ascending=False
    )[["Amount", "Class"]].head(10)
)

# Check genuine vs fraudulent transactions
print("\nOUTLIER CLASS DISTRIBUTION")
print(outliers["Class"].value_counts())

# Statistical summary
print("\nOUTLIER STATISTICS")
display(outliers["Amount"].describe())
```

```python

# DOCUMENT OUTLIER TREATMENT

outlier_decision = {
    "Method": "IQR",
    "Column": "Amount",
    "Potential Outliers": len(outliers),
    "Genuine Transactions": int((outliers["Class"] == 0).sum()),
    "Fraudulent Transactions": int((outliers["Class"] == 1).sum()),
    "Treatment": "Retained",
    "Reason": "Extreme transaction amounts may be legitimate. "
              "Removing or capping them could destroy useful information."
}

print("OUTLIER TREATMENT DOCUMENTATION")
print("=" * 45)

for key, value in outlier_decision.items():
    print(f"{key}: {value}")

print("\nOriginal transaction amounts preserved:",
      df["Amount"].equals(df["Amount"].copy()))
```

```python

# Z-SCORE OUTLIER DETECTION

mean_amount = df["Amount"].mean()
std_amount = df["Amount"].std()

df["Amount_Zscore"] = (
    (df["Amount"] - mean_amount) / std_amount
)

zscore_outliers = df[
    df["Amount_Zscore"].abs() > 3
]

print("Z-SCORE OUTLIER REPORT")
print("=" * 45)

print("Mean Amount:", mean_amount)
print("Standard Deviation:", std_amount)
print("Z-score Threshold: +/- 3")

print("\nPotential outliers:", len(zscore_outliers))

print("Percentage of dataset:",
      round(len(zscore_outliers) / len(df) * 100, 2), "%")

print("\nFraud vs Genuine:")
print(zscore_outliers["Class"].value_counts())
```

```python

# Keep the original dataset free from temporary analysis columns

if "Amount_Zscore" in df.columns:
    df = df.drop(columns=["Amount_Zscore"])

print("Temporary Z-score column removed.")
print("Current number of columns:", df.shape[1])
```

```python

# DATA TYPE VERIFICATION

print("DATA TYPE REPORT")
print("=" * 45)

display(df.dtypes.to_frame(name="Current Data Type"))

print("\nDATA TYPE COUNTS")
print(df.dtypes.value_counts())

print("\nTOTAL COLUMNS:", len(df.columns))

print("\nCHECK IMPORTANT COLUMNS")
print("Time:", df["Time"].dtype)
print("Amount:", df["Amount"].dtype)
print("Class:", df["Class"].dtype)
```

```python

# Verify data types and dataset structure

print("FINAL DATA TYPE CHECK")
print("=" * 40)

checks = {
    "Time is numeric": pd.api.types.is_numeric_dtype(df["Time"]),
    "Amount is numeric": pd.api.types.is_numeric_dtype(df["Amount"]),
    "Class is integer": pd.api.types.is_integer_dtype(df["Class"]),
    "PCA features are numeric": all(
        pd.api.types.is_numeric_dtype(df[col])
        for col in [f"V{i}" for i in range(1, 29)]
    )
}

for check, result in checks.items():
    print(check, ":", "PASS" if result else "FAIL")

print("\nTotal columns:", df.shape[1])
print("Total rows:", df.shape[0])
```
